# Analisi degli LLM scelti per l'esperimento

-- (Llama family) --

## huggingface.co/QuantFactory/Meta-Llama-3.1-8B-Instruct-GGUF:latest
    - https://huggingface.co/bartowski/Meta-Llama-3.1-8B-Instruct-GGUF

    Versione quantizzata in formato GGUF del modello Meta-Llama-3.1-8B-Instruct, ottimizzata per l'esecuzione in locale con basse risorse hardware ed elaborata da QuantFactory.
    Dimensione: 8 miliardi di parametri.
    Finestra di Contesto: Estesa fino a 128k token, ideale per gestire testi e conversazioni molto lunghi.
    Architettura: Transformer autoregressivo con Grouped-Query Attention (GQA) per ottimizzare la velocità di inferenza.
    Addestramento: Ottimizzato tramite Instruction Tuning (SFT e RLHF). Supporta nativamente più lingue tra cui l'italiano, l'inglese, il francese, il tedesco e lo spagnolo.
    Formato GGUF: Permette di eseguire il modello sfruttando sia la CPU che la GPU tramite quantizzazioni di vario livello (es. Q4_K_M, Q8_0), riducendo drasticamente i requisiti di memoria VRAM/RAM rispetto alla versione originale a virgola mobile.

## huggingface.co/QuantFactory/Llama-3.2-3B-Instruct-GGUF:latest 
    - https://huggingface.co/QuantFactory/Llama-3.2-3B-Instruct-GGUF

    Versione quantizzata del modello open-source Llama-3.2-3B-Instruct originariamente sviluppato e rilasciato da Meta.
    Architettura e Parametri: Possiede circa 3,21 miliardi di parametri, utilizza il meccanismo Grouped-Query Attention (GQA) per una maggiore efficienza e supporta una corposa estensione del contesto fino a 128k token.
    Destinazione d'Uso (Instruct): Essendo una variante Instruct, è specificamente ottimizzata per seguire istruzioni in linguaggio naturale, sostenere dialoghi multilingue, riassumere testi e fungere da assistente virtuale.
    Lingue Supportate: È addestrato per comprendere e rispondere in diverse lingue, tra cui italiano, inglese, tedesco, francese, portoghese, spagnolo, hindi e thailandese.

-- (Mistral family) --

## mistral:7b-instruct
    - https://mistral.ai/news/announcing-mistral-7b/

    Modello linguistico open-source da 7,3 miliardi di parametri sviluppato dall'azienda francese Mistral AI. È una versione ottimizzata per seguire istruzioni (fine-tuning di tipo instruction) a partire dal modello di base Mistral-7B. Nonostante le dimensioni ridotte, è celebre per aver ridefinito gli standard di efficienza energetica e computazionale, superando nei benchmark modelli molto più grandi come Llama 2 da 13B.
    Grouped-Query Attention (GQA): Riduce i tempi e i costi di calcolo durante l'inferenza, garantendo risposte più rapide ed efficienti.
    Sliding Window Attention (SWA): Utilizzato nelle prime versioni per gestire sequenze più lunghe riducendo l'uso della cache di memoria.
    Licenza Apache 2.0: Totalmente permissiva, utilizzabile liberamente anche per scopi commerciali ed estremamente facile da personalizzare (fine-tuning).
    Formato Prompt: Gestisce le conversazioni e le istruzioni racchiudendo le richieste dell'utente all'interno dei token speciali [INST] e [/INST].
    
    Il modello è stato aggiornato nel tempo per estenderne le capacità fondamentali:
    Caratteristica	        Versione 0.1	        Versione 0.2	        Versione 0.3 (Ultima)
    Finestra di Contesto	8.000 token	            32.000 token	        32.000 token
    Dimensione Vocabolario	32.000 vocaboli	        32.000 vocaboli	        32.768 vocaboli (miglior supporto multilingua)
    Function Calling (Tool Use)	No	                No	                    Sì (Nativo): Può interfacciarsi con API esterne.
    Tokenizer	            v1	                    v1	                    v3

    Il modello viene distribuito come dimostrazione tecnica e non include meccanismi nativi di moderazione o guardrail. Se impiegato in ambienti di produzione a contatto con il pubblico, è necessario implementare filtri esterni (es. Llama Guard) per evitare la generazione di output tossici o inappropriati.

## mistral-nemo:12b
    - https://mistral.ai/it/news/mistral-nemo/

    Modello linguistico di grandi dimensioni da 12 miliardi di parametri, sviluppato in collaborazione da Mistral AI e NVIDIA e rilasciato a luglio 2024 con licenza open-source Apache 2.0. Pensato come sostituto diretto del precedente Mistral 7B, si posiziona come un modello di categoria "enterprise" ottimizzato per offrire prestazioni elevate su hardware accessibile, come una singola GPU.
    Finestra di contesto: Fino a 128k token.
    Architettura: Standard Transformer basato su Llama con 40 livelli, 32 attention heads e Grouped-Query Attention (GQA).
    Supporto Nativo FP8: Addestrato con consapevolezza della quantizzazione (quantization-aware training), permette l'inferenza in formato FP8 riducendo drasticamente l'uso della memoria RAM/VRAM senza alcuna perdita di accuratezza.
    Tokenizzatore "Tekken": Un nuovo tokenizzatore basato su Tiktoken con un vocabolario di 128k o 131k token. Risulta circa il 30% più efficiente nella compressione del codice sorgente e dei testi in numerose lingue (tra cui l'italiano) rispetto al vecchio SentencePiece.
    
    Multilinguismo avanzato: Prestazioni eccellenti in 11 lingue principali (Inglese, Italiano, Francese, Tedesco, Spagnolo, Portoghese, Cinese, Giapponese, Coreano, Arabo e Hindi).
    Funzionalità ed Esecuzione: Ottime capacità di ragionamento logico, generazione e correzione di codice di programmazione e supporto nativo per il function calling (chiamate a strumenti esterni).
    Disponibilità: Rilasciato sia in versione pre-addestrata (Base) che ottimizzata per i dialoghi (Instruct). È facilmente integrabile tramite piattaforme come Hugging Face, Ollama (eseguibile tramite il comando ollama run mistral-nemo:12b), ed è distribuito come microservizio ottimizzato NVIDIA NIM.

    Nei test standard per la sua fascia di dimensioni (modelli sotto i 15B), supera o compete direttamente con rivali come Llama 3 8B e Gemma 2 9B:
    MMLU (Conoscenza generale): ~68.0%
    HellaSwag (Ragionamento): ~83.5%
    TriviaQA (Domande/Risposte): ~73.8%

-- (Qwen family) --

## qwen2.5:7b-instruct
    - https://huggingface.co/Qwen/Qwen2.5-7B-Instruct

    Uno dei modelli linguistici open-weight più apprezzati della famiglia Qwen 2.5 sviluppata da Alibaba Cloud
    Architettura e Dimensione: È un modello generativo basato su architettura Transformer con 7 miliardi di parametri, ottimizzato per seguire istruzioni (Instruction-tuned).
    Finestra di Contesto: Supporta una lunghezza di contesto fino a 128.000 token, ideale per elaborare testi molto lunghi o intere conversazioni.
    Capacità Multilingue: Offre prestazioni eccellenti in oltre 29 lingue, tra cui l'italiano, mostrando una forte comprensione delle sfumature culturali e linguistiche.
    Punti di Forza Tecnici: Eccelle particolarmente nella scrittura di codice (coding), nella risoluzione di problemi matematici e nel ragionamento logico rispetto ad altri modelli di pari dimensioni.
    Efficienza: Grazie alla sua dimensione compatta (7B), è ampiamente utilizzato per il deployment locale (tramite tool come Ollama) su normali computer consumer dotati di una buona GPU.


## qwen2.5:14b-instruct
    - https://ollama.com/library/qwen2.5:14b-instruct

    Modello di linguaggio open-weight sviluppato da Alibaba Cloud, ottimizzato per l'istruzione e il dialogo (instruction-tuned). Rappresenta uno dei compromessi ideali tra prestazioni ed efficienza energetica all'interno dell'ecosistema open-source, integrandosi nativamente in piattaforme di esecuzione locale come Ollama.
    Parametri totali: 14,7 miliardi (13,1 miliardi non-embedding).
    Finestra di contesto: Supporta fino a 128K token in input (fino a 131.072 con scaling YaRN) e genera un output massimo di 8K token.
    Dataset di addestramento: Pre-addestrato su ben 18 trilioni di token.
    Architettura: Transformer di tipo decoder-only denso, dotato di Grouped-Query Attention (GQA) e attivazione SwiGLU.
    Licenza: Apache 2.0, che ne permette un ampio utilizzo commerciale e di ricerca.
    
    Programmazione e Matematica: Ha ricevuto un netto potenziamento nel ragionamento logico, nella scrittura/correzione di codice e nella risoluzione di problemi matematici complessi.
    Output Strutturati: Eccelle nella comprensione di dati tabellari e nella generazione di risposte in formato strutturato, come schemi JSON puliti.
    Multilinguismo: Supporta nativamente oltre 29 lingue, dimostrando ottime proprietà di localizzazione e traduzione anche per l'italiano.
    Flessibilità dei Prompt: Risulta molto più resiliente alle variazioni nei system prompt, facilitando l'impostazione di giochi di ruolo (role-play) o la configurazione di scenari specifici per chatbot.

    I requisiti variano molto in base alla quantizzazione (il file immagine scaricabile):
    Quantizzazione standard Ollama (Q4_K_M): Richiede un download di circa 9.0 GB. Gira fluidamente su GPU consumer dotate di almeno 12-16 GB di VRAM (es. RTX 3060/4060 o superiori).
    Precisione FP16 (non quantizzato): Richiede circa 34 GB di VRAM per l'allocazione base del modello e del contesto iniziale.
